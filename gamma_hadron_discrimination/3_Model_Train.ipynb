The model is stored in a directory containing all .joblil files and the plot importances values, to be called later on. A snippet of code is used to show number of instances for each class and their relative ratio (baseline for PRAUC).

- Specify model and feature importances save directory (should be the same!).
- Set minimum and maximum distances for the stations and minimum pe number per station.
