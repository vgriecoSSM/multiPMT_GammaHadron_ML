```python
import sys
sys.path.append("../")
import uproot
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
from py_code import DATroot_primary_reader as dat
```

```python
dat_path_gammas = "/swgo/Array_Simulato_Analisi_ML/roots_mPMT_gammas_SWGO_array/4FF_R600m_h175r180_black_uniform0600m_100300TeV_030deg/"
min_dat = 0
max_dat = 1
max_time = 600
```

```python
import uproot

for dat_idx in range(min_dat, max_dat):

    DAT_name = f"DAT10{dat_idx:04d}"
    DAT_file_name = f"{DAT_name}.root"
    DAT_path = dat_path_gammas + DAT_file_name

    with uproot.open(DAT_path) as root_file:
        tree = root_file["XCDF"]
        df = tree.arrays(library="pd")

    evts = df["HAWCSim.Evt.Num"].unique()
    evts_number = len(evts)

    for event_id in evts:

        p = df[df["HAWCSim.Evt.Num"] == event_id].copy()

        p = dat.get_stations_info(
            p,
            "../survey_and_array_txt_repo/tank_pos_H_4FF.txt"
        )

        if p is None:
            continue

        p = dat.get_shower_time(p)

        check_trigger_time = p["Trigger_Time"].iloc[0]

        if np.isnan(check_trigger_time.all()):
            continue

        p = dat.get_dt_correction(p)
        p = dat.appy_time_correction(p)
        p_test = dat.get_st_lvl_test_df(p, max_time=max_time)

        pe_r_tank = p_test["PE_RTank"].to_numpy()

        print(len(pe_r_tank))
```
