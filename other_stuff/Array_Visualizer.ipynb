```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("../")
path_txt_tank_positions = "../survey_and_array_txt_repo/tank_pos_H_4FF.txt"
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from py_code import add_variables
from py_code import get_variables
from py_code import initialize_df
import matplotlib
matplotlib.rc('text', usetex=True)
matplotlib.rcParams['text.latex.preamble'] = r"\usepackage{amsmath,amssymb}"
plt.style.use('plots.mplstyle')
from matplotlib.patches import Patch, Rectangle
```

```python
idxch, xch, ych, rch = np.loadtxt("../survey_and_array_txt_repo/tank_pos_H_4FF.txt", delimiter = " ", unpack = True)
xt = xch.reshape(-1, 7)
yt = ych.reshape(-1, 7)

print("Number of Tanks = ", len(xt))
print("Number of Tanks = ", len(yt))
```

```python
xt_m = xt/100
yt_m = yt/100

fig, ax = plt.subplots(figsize=(8, 8))

ax.scatter(xt_m, yt_m, s=15, color="royalblue")

ax.set_xlabel("x [m]")
ax.set_ylabel("y [m]")
ax.set_title("Tank array")
ax.set_aspect("equal", adjustable="box")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import random

folder = Path("../output_files_ML/mPMT_HAWCSim_array/test_protons/100300TeV_030deg_4FF_31381channels_180R175h_Black/NoneMHz_ADC_13petrain_13petest_500ns_0p2nsTDC/test_dist_0000_0700m_13pe_65multiplicity")

file = random.choice(list(folder.glob("*.txt")))
print(file.name)

truelabel, pmu, tc, nomtheta, nomen, yc, xc, tr, tx, ty = np.loadtxt(file, unpack=True)
```

```python
Etev = nomen/1000
```

```python
fig, ax = plt.subplots(figsize=(16, 10))

true_mutanks_x = tx[truelabel == 1]
true_mutanks_y = ty[truelabel == 1]

ax.scatter(xt_m, yt_m, s=15, color="lightgray", zorder=1)
ax.scatter(true_mutanks_x, true_mutanks_y, marker="o", s=15, facecolors="none", edgecolors="red", linewidths=1, label=r"stations w/ $\mu^{\pm}$", zorder=3)

sc = ax.scatter(tx, ty, c=pmu, cmap="viridis", vmin=0, vmax=1, s=15, zorder=2)

cbar = fig.colorbar(sc, ax=ax, orientation="horizontal", pad=0.12, fraction=0.05)
cbar.set_label(r"$\mathrm{P}_{\mu}$", fontsize=22)
cbar.ax.tick_params(labelsize=22, direction="in", length=6, width=1.2)
cbar.outline.set_linewidth(1.2)

ax.set_title(rf"Proton" "\n" rf"$\mathrm{{E}} = {np.atleast_1d(Etev)[0]:.2f}$ TeV" "\n" rf"$\theta = {np.atleast_1d(nomtheta)[0]:.2f}^{{\circ}}$", fontsize=18)
ax.set_xlabel(r"$\mathrm{x\ [m]}$", fontsize=24)
ax.set_ylabel(r"$\mathrm{y\ [m]}$", fontsize=24)
ax.set_xlim(-640, 640)
ax.set_ylim(-640, 640)

ax.set_aspect("equal", adjustable="box")
ax.minorticks_on()
ax.tick_params(axis="both", which="major", direction="in", top=True, right=True, labelsize=20, length=7, width=1.2)
ax.tick_params(axis="both", which="minor", direction="in", top=True, right=True, length=4, width=1)

for spine in ax.spines.values():
    spine.set_linewidth(1.2)

ax.grid(alpha=0.2)

legend = ax.legend(fontsize=12, frameon=True, edgecolor="black", facecolor="white", framealpha=1, loc=[0.73,0.92])
legend.get_frame().set_linewidth(1.2)

plt.tight_layout()
plt.show()
```
